```python
import pandas as pd

file_path = "../data/raw/household_power_consumption.txt"

df = pd.read_csv(
    file_path,
    sep=";",
    na_values="?"
    
)

df.head()
print(df.head())
print(df.tail())
print(df.shape)
print(df.info())
print(df.isna().sum())
```

```python
# Exploring one column

print(df["Global_active_power"].head())
print(df["Global_active_power"].max())
print(df["Global_active_power"].mean())
print(df["Global_active_power"].min())
print(df["Global_active_power"].median())
```

```python
# Finding high power readings

high_power = df[df["Global_active_power"] > 5]
print(high_power.head())
print(len(high_power))
print(high_power.shape[0])
```

```python
# Finding missing data
#something[something > condition]

missing = df.isna().sum()
print(missing[missing > 0])
```

```python
# Creating timestamp (for making time-series analysis easier)
#This will be essential for:
#- missing intervals
#- duplicate timestamps
#- hourly aggregation
#- daily aggregation
#- anomaly detection
#- time-based SQL queries

df["timestamp"] = pd.to_datetime(
    df["Date"] + " " + df["Time"],
    dayfirst=True
)

df[["Date", "Time", "timestamp"]].head()
```

```python
# Extract information from timestamp

df["timestamp"].dt.year
```

```python
df["timestamp"].dt.month
#df["timestamp"].dt.day
#df["timestamp"].dt.hour
df["hour"] = df["timestamp"].dt.hour
df[["timestamp", "hour"]].head()
```

```python
# Create Weekday Information
df["day_of_week"] = df["timestamp"].dt.day_name()
df[["timestamp", "day_of_week"]].head()
```

```python
# Grouping

df.groupby("day_of_week")["Global_active_power"].max()
#df.groupby("hour")["Global_active_power"].median()
```

```python
# Group by hour
# basic Energy analysis
hourly_average = (
    df.groupby("hour")["Global_active_power"].mean()
)

print(hourly_average)
```

```python
# Resampling

# making timestamp as index as we have the value
df = df.set_index("timestamp")
print(df.head())
```

```python
# Calculating daily/hourly average

daily_power = df["Global_active_power"].resample("D").mean()
#hourly_power = df["Global_active_power"].resample("H").mean()
print(daily_power)
```

```python
# how many rows/columns
rows, columns = df.shape
print(f"Rows: {rows}, Columns: {columns}")
# column names
print(df.columns)
# missing value in columns
print(df.isnull().sum())
# Average global_active_power
print(df["Global_active_power"].mean())
# Max
print(df["Global_active_power"].max())
# All records with condition
record5 = df[df["Global_active_power"] > 5 ]
print(record5)
# Create timstamp column
df["timestamp_1"] = pd.to_datetime(
    df["Date"] + " - " + df["Time"],
    dayfirst=True
)
print(df["timestamp_1"])
```

```python
# Challange

def create_data_quality_summary(df):

    total_rows = df.shape[0]
    total_columns = df.shape[1]

    missing_values = df.isna().sum().sum()

    duplicate_rows = df.duplicated().sum()

    result = {
        "total_rows": total_rows,
        "total_columns": total_columns,
        "missing_values": missing_values,
        "duplicate_rows": duplicate_rows
    }

    return result

result = create_data_quality_summary(df)

print(result)
```
